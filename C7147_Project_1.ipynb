## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
def validate_read(sequence, reference):
    """Validate an individual read against a reference sequence."""
    # converting the sequence into uppercase
    normalised_seq = sequence.upper()

    errors = [] # if there are no errors across the checks then the errors remain []

    # Check that the read has the correct length
    if len(normalised_seq) != len(reference):
        errors.append("wrong_length")

    # Check for any invalid characters
    valid_bases = {"A", "C", "G", "T"}
    if not all(base in valid_bases for base in normalised_seq):
        errors.append("invalid_characters")

    # Determine validity based on whether errors exist
    is_valid = len(errors) == 0 # It is only True if it passes both checks

    return {
        "sequence": normalised_seq,
        "valid": is_valid,
        "errors": errors
    }
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


### 💻 Enter your code here

````python
# return a list containing one dictionary per mismatch in ascending order
def find_mutations(sequence, reference):
    """Identifies differences between a valid and reference sequence."""
    mutations = [] # if no mutations are found it will return []

    for i in range(len(sequence)):
        if sequence[i] != reference[i]:
            mutations.append({
                "position": i + 1, # position of mutation
                "reference_base": reference[i], # one-character string from reference
                "read_base": sequence[i] # one-character string from sequence
            })
    return mutations
````

### 💣 Test your code

````python
# tests for step 2

# normal case with multiple mismatches
test_mutations = find_mutations("GTGA", "ATGC")
assert test_mutations == [
    {"position": 1, "reference_base": "A", "read_base": "G"},
    {"position": 4, "reference_base": "C", "read_base": "A"}]
print(test_mutations)

# edge case, zero mutations
exact_match = find_mutations("ATGC", "ATGC")
assert exact_match == []
print(exact_match)

# invalid, mismatch at first and last position
test_edge = find_mutations("GTGA", "ATGC")
assert test_edge[0]["position"] == 1
assert test_edge[-1]["position"] == 4
print(test_edge)
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:
- `reads` is a list of dictionaries containing keys `read_id`, `sample`, `group` and `sequence`, each with string values.
- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python

# combining validate_read and find_mutations to create dictionaries
def analyse_reads(reads, reference):
    """Returns a new list of dictionaries with status, errors, and mutations for each read."""
    analysed_list = []

    for r in reads:
        # validate the current sequence
        validation = validate_read(r["sequence"], reference)

        if validation["valid"]:
            status = "valid"
        else:
            status = "invalid"

        errors = validation["errors"]

        # find mutations only if the read is valid
        if status == "valid":
            mutations = find_mutations(validation["sequence"], reference)
        else:
            mutations = []

        # making each dictionary use the keys given above
        analysed_read = {
            "read_id": r["read_id"],
            "sample": r["sample"],
            "group": r["group"],
            "sequence": validation["sequence"],
            "status": status,
            "errors": errors,
            "mutations": mutations
        }

        analysed_list.append(analysed_read)

    return analysed_list
````

## Step 4 — Summarise mismatch patterns between samples

## Function: `summarise_samples(analysed_reads)`

### 👍 You can assume:

### 🎯 Aims:

- Use your output from the `analyse_reads` function (Step 3) to develop a sample-level analysis that could help a researcher investigate differences in mismatch events between one or more groups (e.g. genotoxic agent-treated vs. control in the example dataset).

- Calculate percentages using **valid reads only**. If a sample has no valid reads, clearly indicate that its percentage cannot be calculated.

**Table of summarised results**

    - The three positions with the most mismatch events (or fewer if less than 3 are found), selecting the earlier position(s) in the sequence in the case of a tie.

Return the results in the following structure, with one nested dictionary per sample, e.g. (this is an example of the format, not a correct numerical output):

### 💻 Enter your code here

````python
# using output from the analysed_reads to develop sample-level analysis investigating differences in mismatch events


def summarise_samples(analysed_reads):
  """Summarise mismatch patterns between samples."""
  sample_summary = {}

  # aggregating raw counts by sample
  for r in analysed_reads:
    sample = r["sample"]

    if sample not in sample_summary:
      sample_summary[sample] = {
          "group": r["group"],
          "valid_reads": 0,
          "invalid_reads": 0,
          "mismatch_reads": 0,
          "mismatch_percentage": None,
          "mismatch_events": 0,
          "substitution_counts": {},
          "position_counts": {},
          "top_positions": []
      }

    stats = sample_summary[sample]
    if r["status"] == "valid":
      stats["valid_reads"] += 1
      if len(r["mutations"]) > 0:
        stats["mismatch_reads"] += 1
        stats["mismatch_events"] += len(r["mutations"])
        for mutation in r["mutations"]:
          position = mutation["position"]
          if position not in stats["position_counts"]:
            stats["position_counts"][position] = 0
          stats["position_counts"][position] += 1
          # Tracking for the substitution counts
          sub_key = f"{mutation['reference_base']}>{mutation['read_base']}"
          if sub_key not in stats["substitution_counts"]:
            stats["substitution_counts"][sub_key] = 0
          stats["substitution_counts"][sub_key] += 1
    else:
      stats["invalid_reads"] += 1

  # calculate percentages using valid reads
  for sample, stats in sample_summary.items():
    if stats["valid_reads"] > 0:
      stats["mismatch_percentage"] = (
          stats["mismatch_reads"] / stats["valid_reads"]) * 100
    else:
      stats["mismatch_percentage"] = None # indicates inability of calculation

    # determine the top 3 positions with the most mismatch events
    # smaller position number first
    pos_counts = stats["position_counts"]
    sorted_positions = sorted(
        pos_counts.keys(), key=lambda p: (-pos_counts[p], p))
    stats["top_positions"] = sorted_positions[:3]

  # print a summary section for each group separately using data from each sample
  groups = {}
  for sample, stats in sample_summary.items():
    g = stats["group"]
    if g not in groups:
      groups[g] = {}
    groups[g][sample] = stats

  for group_name, samples in sorted(groups.items()): # labels the individual groups
    print(f"\n\nSummary Table for {group_name}\n{'-' * 85}") # adds --- line
    print(f"{'Sample' :<10} | {'Valid':<8} | {'Invalid':<8} | {'Mismatch Read %':<15} | {'Mismatch Events':<15} | {'Top Positions'}")
    # the :<number allow the table to be more evenly spaced, with gaps between the |
    print("-" * 85)
    for sample, stats in sorted(samples.items()):
      mismatch_pct = stats["mismatch_percentage"]
      mismatch_pct_str = f"{mismatch_pct:.2f}%" if mismatch_pct is not None else "N/A"
      top_pos_str = ", ".join(map(str, stats["top_positions"]))
      print(f"{sample:<10} | {stats['valid_reads']:<8} | {stats['invalid_reads']:<8} | {mismatch_pct_str:<15} | {stats['mismatch_events']:<15} | {top_pos_str}")

    # print substitution table
    all_substitutions = set()
    for sample, stats in samples.items():
      all_substitutions.update(stats["substitution_counts"].keys())
    sorted_substitutions = sorted(list(all_substitutions))

    print(f"\n\nSubstitution Table for {group_name}") # title and adding gaps between the tables
    print("-" * 85)
    samples_in_group = sorted(samples.keys())
    header = f"{'Substitution' :<15}"
    for sample in samples_in_group:
      header += f" | {sample:<20}"
    print(header)
    print("-" * 85)

    for substitution in sorted_substitutions:
      row = f"{substitution :<15}"
      for sample in samples_in_group:
        stats = samples[sample]
        count = stats["substitution_counts"].get(substitution, 0)
        val_reads = stats["valid_reads"]
        pct_val_reads = (count / val_reads * 100) if val_reads > 0 else 0
        row_item = f"{count:<5} ({pct_val_reads:.2f}%)"
        row += f" | {row_item:<20}" # This just more evenly spaces the columns
      print(row)

  return sample_summary


# analyse all loaded reads using step 3
analysed_dataset = analyse_reads(reads, reference)

# generate and print tables using step 4
results = summarise_samples(analysed_dataset)
````

## Step 5 — Exploratory comparison of the top mismatch positions

## Function:
`compare_positions(sample_summary, control_group="control", treated_group="treatment")`

### 👍 You can assume

- `sample_summary` is a dictionary in the format returned by the `summarise_samples` function
- `control_group` and `treated_group` are strings
- All samples have been analysed against the same reference sequence.

### 🎯 Aims

- Using the `position_counts` returned by `summarise_samples`, sum the mismatch counts at every position, **combining samples from `control_group` and `treatment_group`** (you should end up with a 1D array/list with the same length as a read).
- Select the **top 3 positions with the highest combined counts** (or fewer if there are less than three mismatches), choosing the earlier reference position in the case of a tie.
- For each position selected, calculate **mismatch percentages** separately for each biological sample.
- Compare treatment groups using the supplied statistical function `calculate_statistics` (see below).
- **Print out** the results of the statistical comparison clearly in standard scientific style, including appropriate statistical details based on the test used (a two-sided Welch's t-test).

🛠 **Parameters:**

- `sample_summary`: output from `summarise_samples` function.
- `control_group` and `treated_group`: labels identifying the groups to compare.

### 🧮 Sample-level calculations

For each selected position and sample, calculate:

**Mismatch percentage = valid reads with a mismatch at that position ÷ total valid reads × 100**

Count any substitution at the selected position. Include samples with zero mismatches, but exclude samples with no valid reads and report their exclusion.

Reuse helper functions from Step 4 where appropriate to reduce code complexity and repetition.

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````

### 📋 Return format

Return a list with one dictionary per selected position:

```python
[
    {
        "position": 42,
        "control_n": 3,
        "treated_n": 3,
        "control_mean": 0.5,
        "treated_mean": 2.0,
        "difference": 1.5,
        "statistic": ...,
        "p_value": ...
    }
]
```

Means are sample mismatch percentages; `difference` is **treated mean − control mean**, in percentage points. Use `None` for results that cannot be calculated.

Print **one brief line per position**, reporting group means, their difference, sample numbers and p-value.

````python
# sum the mismatch counts combining samples from control and treatment
def compare_positions(sample_summary,
                      control_group="control",
                      treated_group="treatment"):
  """Summarise mismatch patterns at every position."""
  if not sample_summary:
    return []

  total_count = {}
  for sample, stats in sample_summary.items():
    if stats["group"] in (control_group, treated_group):
      for position, count in stats["position_counts"].items():
        total_count[position] = total_count.get(position, 0) + count

  # select the top 3 positions with the highest combined counts - earlier reference first
  sorted_positions = sorted(total_count.keys(),
                            key=lambda p: (-total_count[p], p))
  top_positions = sorted_positions[:3]
  compare_results = []

# printing table header outside of the following loop - no longer needed as simple line
  # print(f"\nStatistical Comparison of Top Mismatch Positions")
  # print("-" * 105)
  # print(f"{'Position' :<10} | {'Control Mean' :<14} | {'Treated Mean' :<14} | {'Difference' :<12} | {'Samples' :<14} | {'t-stat' :<10} | {'p-value' :<10}")
  # print("-" * 105)
  # no longer needed as it is supposed to be a simple line

  # calculate mismatch percentages separately for each bio sample
  for position in top_positions:
    control_percentages = []
    treated_percentages = []
    control_used = 0
    treated_used = 0

    for sample, stats in sample_summary.items():
      group = stats["group"]
      valid_reads = stats["valid_reads"]

      if group == control_group:
        if valid_reads > 0:
          stat_counts = stats["position_counts"].get(position, 0)
          control_percentages.append(stat_counts / valid_reads * 100)
          control_used += 1
      elif group == treated_group:
        if valid_reads > 0:
          stat_counts = stats["position_counts"].get(position, 0)
          treated_percentages.append(stat_counts / valid_reads * 100)
          treated_used += 1

    # compare treatment groups using calculate_statistics
    stat_results = calculate_statistics(control_percentages, treated_percentages)

# print results of statistical comparison
    # calculating means and differences
    control_mean = (sum(control_percentages) / control_used
                    if control_used > 0
                    else None)
    treated_mean = (sum(treated_percentages) / treated_used
                    if treated_used > 0
                    else None)
    difference = (treated_mean - control_mean
                  if treated_mean is not None and control_mean is not None
                  else None)
    compare_results.append({
    "position": position,
    "control_n": control_used,
    "treated_n": treated_used,
    "control_mean": control_mean,
    "treated_mean": treated_mean,
    "difference": difference,
    "statistic": stat_results["statistic"],
    "p_value": stat_results["p_value"]
})
    # samples = f"{control_used} / {treated_used}"

    # print(f"{position:<10d} | {control_mean:<14.2f} | {treated_mean:<14.2f} | {difference:<12.2f} | {samples:<14} | {stat_results['statistic']:<10.4f} | {stat_results['p_value']:<10.4e}")
    # no longer needed as brief lines wanted
    print(f"Position {position}: Control = {control_mean:4f} (n = {control_used}), Treatment = {treated_mean:4f} (n = {treated_used}), Difference = {difference:4f} p-value:{stat_results['p_value']:4e}")


  return compare_results

# printing

analysis_results = compare_positions(results, control_group="control", treated_group="treatment")
````
